## matplotlib Basics

```python
import matplotlib.pyplot as plt
```

```python
# Einfachste Version

plt.plot(df["Monat"], df["Durchschnittlicher Tagesverkehr nach Monat"])
plt.xlabel("Monat")
plt.ylabel("Durschnittler Verkehr")
plt.title("Verkehr in Vintl im Jahr 2024")
plt.grid(True)
plt.show()
```

```python
# Mehr Kontrolle:

fig, ax = plt.subplots(figsize=(10, 4))

ax.plot (df["Monat"], df["Durchschnittlicher Tagesverkehr nach Monat"], label = "Durschnittlicher Verkehr")
ax.set_xlabel("Monat")
ax.set_ylabel("Durschnittlicher Verkehr")
ax.legend()
```

```python
# Bild speichern

fig.savefig("Graphik.png", dpi=150)

fig.savefig("Graphik.pdf")
```

## plotly Basics

```python
import plotly.express as px
```

```python
fig = px.line (
    df,
    x = "Monat",
    y = "Durchschnittlicher Tagesverkehr nach Monat",
    title = "Vintl 2024",
)

fig.show()
```

```python
fig = px.line (
    df,
    x = "Monat",
    y= ["Durchschnittlicher Tagesverkehr nach Monat", "Leicht", "Schwer"],
    title = "Vintl 2024"
)

fig.show()
```
